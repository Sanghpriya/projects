```python
import rasterio
import numpy as np

input_path = r"D:\landslide\dem_derivatives\TWI.tif"
output_path = r"D:\landslide\dem_derivatives\TWI_clean.tif"

with rasterio.open(input_path) as src:
    arr = src.read(1).astype("float32")
    meta = src.meta.copy()

    nodata = src.nodata
    print("Declared nodata:", nodata)

    arr[arr == nodata] = np.nan

    meta.update(dtype="float32", nodata=np.nan)

    with rasterio.open(output_path, "w", **meta) as dst:
        dst.write(arr, 1)

print("TWI cleaned.")
```

```python
import rasterio
from rasterio.warp import reproject, Resampling
import numpy as np
import os

# -----------------------------
# Reference grid (10 m master)
# -----------------------------
reference = r"D:\landslide\final_data\distance_roads_tehri.tif"

with rasterio.open(reference) as ref:
    ref_meta = ref.meta.copy()
    ref_transform = ref.transform
    ref_crs = ref.crs
    ref_shape = (ref.height, ref.width)

# -----------------------------
# Continuous rasters (bilinear)
# -----------------------------
continuous_layers = [
    r"D:\landslide\final_data\Slope_deg.tif",
    r"D:\landslide\final_data\Aspect_deg.tif",
    r"D:\landslide\final_data\PlanCurv.tif",
    r"D:\landslide\final_data\ProfileCurv.tif",
    r"D:\landslide\final_data\TRI.tif",
    r"D:\landslide\final_data\TWI_clean.tif",
    r"D:\landslide\final_data\Rainfall_Tehri_clean.tif",
]

# -----------------------------
# Categorical rasters (nearest)
# -----------------------------
categorical_layers = [
    r"D:\landslide\final_data\Soil_Tehri.tif",
    r"D:\landslide\final_data\Geomorphon_clean.tif",
    r"D:\landslide\final_data\lulc_tehri.tif",
]

# -----------------------------
# Function to resample
# -----------------------------
def resample_raster(input_path, resampling_method, dtype, nodata_value):
    output_path = input_path.replace(".tif", "_10m.tif")

    with rasterio.open(input_path) as src:
        data = src.read(1)
        resampled = np.empty(ref_shape, dtype=dtype)

        reproject(
            source=data,
            destination=resampled,
            src_transform=src.transform,
            src_crs=src.crs,
            dst_transform=ref_transform,
            dst_crs=ref_crs,
            resampling=resampling_method
        )

    new_meta = ref_meta.copy()
    new_meta.update(dtype=dtype, nodata=nodata_value)

    with rasterio.open(output_path, "w", **new_meta) as dst:
        dst.write(resampled, 1)

    print(f"Resampled: {os.path.basename(output_path)}")


# -----------------------------
# Run continuous (bilinear)
# -----------------------------
for layer in continuous_layers:
    resample_raster(
        layer,
        Resampling.bilinear,
        np.float32,
        np.nan
    )

# -----------------------------
# Run categorical (nearest)
# -----------------------------
for layer in categorical_layers:
    resample_raster(
        layer,
        Resampling.nearest,
        np.int16,
        -9999
    )

print("All rasters resampled to 10 m successfully.")
```

```python
import rasterio
from rasterio.warp import reproject, Resampling
import numpy as np
import os

# 1. Load the 10m Master Reference (The grid all files must match)
reference_path = r"D:\landslide\final_data\distance_roads_tehri.tif"

with rasterio.open(reference_path) as ref:
    ref_meta = ref.meta.copy()
    ref_transform = ref.transform
    ref_crs = ref.crs
    ref_shape = (ref.height, ref.width)

# 2. Define Input Files and their specific resampling methods
# Curvature -> Bilinear (Continuous terrain)
# Aspect -> Nearest (Directional degrees)
layers = {
    r"D:\landslide\dem_derivatives\Curvature.tif": Resampling.bilinear,
    r"D:\landslide\dem_derivatives\Aspect_deg.tif": Resampling.nearest
}

def resample_to_new_10m(input_path, method):
    # This creates a NEW filename (e.g., Aspect_deg_10m.tif)
    output_path = input_path.replace(".tif", "_10m.tif")
    
    with rasterio.open(input_path) as src:
        # We use float32 for high-precision terrain decimals
        data = src.read(1).astype(np.float32)
        
        # Standardize NoData to -9999 for ML (Scikit-learn/XGBoost)
        nodata_val = -9999.0
        resampled = np.full(ref_shape, nodata_val, dtype=np.float32)

        reproject(
            source=data,
            destination=resampled,
            src_transform=src.transform,
            src_crs=src.crs,
            dst_transform=ref_transform,
            dst_crs=ref_crs,
            resampling=method,
            src_nodata=src.nodata,  # Detects original NoData (e.g., 0 or -32768)
            dst_nodata=nodata_val   # Writes the new consistent -9999.0
        )

    # Prepare metadata for the NEW file based on the 10m Master
    new_meta = ref_meta.copy()
    new_meta.update(dtype=np.float32, nodata=nodata_val)

    with rasterio.open(output_path, "w", **new_meta) as dst:
        dst.write(resampled, 1)
    
    print(f"Created New File: {os.path.basename(output_path)} using {method.name}")

# Run the process
for path, method in layers.items():
    if os.path.exists(path):
        resample_to_new_10m(path, method)
    else:
        print(f"Error: Could not find {path}")

print("\nStandardization complete. You now have new 10m versions of Aspect and Curvature.")
```
